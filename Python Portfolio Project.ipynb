# Vehicle Fuel Economy Analysis

## Project Objective

The objective of this project is to analyze vehicle fuel economy using Python, Pandas, Seaborn, and Matplotlib.

This analysis explores five questions:

1. How does horsepower relate to fuel efficiency?
2. Which vehicle origin has the highest average MPG?
3. How has fuel economy changed over time?
4. How does vehicle weight affect fuel efficiency?
5. How does cylinder count relate to fuel economy?

## Dataset

The analysis uses Seaborn's MPG dataset, containing vehicle characteristics such as horsepower, weight, cylinders, model year, origin, and miles per gallon.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

mpg = sns.load_dataset("mpg")

mpg.head()
```

## 1. Horsepower vs. Fuel Economy

```python
sns.scatterplot(data=mpg, x="horsepower", y="mpg")
plt.title("Horsepower vs. Fuel Economy")
plt.show()
```

### Key Finding

**Negative correlation:** As horsepower increases, fuel economy generally decreases.

**Interpretation:** Higher-horsepower vehicles in this dataset tend to achieve fewer miles per gallon.
    

## 2. Average Fuel Economy by Country of Origin

```python
mpg.groupby("origin")["mpg"].agg(["count", "mean"]).round(2)
```

### Key Finding

**Regional differences:** Japanese vehicles have the highest average fuel economy in the dataset, followed by European and American vehicles.

**Interpretation:** Vehicle origin appears to be associated with fuel efficiency, although differences in vehicle size, weight, and engine characteristics may also contribute.

## 3. Fuel Economy Trends Over Time

```python
sns.lineplot(data=mpg, x="model_year", y="mpg", errorbar=None)
plt.title("Average Fuel Economy Over Time")
plt.xlabel("Model Year")
plt.ylabel("Average MPG")
plt.show()
```

### Key Finding

**Overall upward trend:** Average fuel economy generally improved across model years, although there were noticeable year-to-year fluctuations.

**Interpretation:** Vehicle fuel efficiency improved over the period represented in the dataset, but progress was not consistent every year.

```python
sns.scatterplot(data=mpg, x="weight", y="mpg")
plt.title("Vehicle Weight vs. Fuel Economy")
plt.xlabel("Vehicle Weight (lbs)")
plt.ylabel("Miles Per Gallon (MPG)")
plt.show()
```

### Key Finding

**Strong negative relationship:** As vehicle weight increases, fuel economy generally decreases.

**Interpretation:** Heavier vehicles tend to achieve fewer miles per gallon, suggesting that vehicle weight is an important factor associated with fuel efficiency.

## 5. Average Fuel Economy by Cylinder Count

```python
sns.barplot(data=mpg, x="cylinders", y="mpg", errorbar=None)
plt.title("Average Fuel Economy by Cylinder Count")
plt.xlabel("Number of Cylinders")
plt.ylabel("Average MPG")
plt.show()
```

### Key Finding

**Significant efficiency differences:** Eight-cylinder vehicles have substantially lower average fuel economy compared to vehicles with fewer cylinders.

**Unexpected result:** Three-cylinder vehicles do not outperform four-cylinder vehicles in this dataset, despite having fewer cylinders.

**Interpretation:** Cylinder count alone does not determine fuel efficiency. Other factors, including vehicle weight, horsepower, and engine design, may contribute to these differences.

## Conclusion

This analysis examined the relationships between vehicle characteristics and fuel economy using Python, Pandas, Seaborn, and Matplotlib.

### Key Findings

- **Horsepower:** Vehicles with higher horsepower generally have lower fuel efficiency.
- **Country of Origin:** Japanese vehicles had the highest average MPG in the dataset.
- **Model Year:** Fuel economy generally improved over time, although there were noticeable fluctuations.
- **Vehicle Weight:** Heavier vehicles consistently demonstrated lower fuel efficiency.
- **Cylinder Count:** Eight-cylinder vehicles had substantially lower average MPG, while three-cylinder vehicles did not outperform four-cylinder vehicles.

### Final Takeaway

Vehicle weight, horsepower, and cylinder count all show relationships with fuel efficiency. However, no single characteristic fully explains fuel economy.

The analysis also demonstrates the importance of examining unexpected results rather than relying entirely on initial assumptions.
